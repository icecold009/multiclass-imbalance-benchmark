```python
# Read-only remote preflight: no data access and no benchmark execution.
from pathlib import Path
import os
import platform
import shutil
import sys

def memory_gib():
    if hasattr(os, 'sysconf'):
        pages = os.sysconf('SC_PHYS_PAGES')
        size = os.sysconf('SC_PAGE_SIZE')
        return pages * size / (1024 ** 3)
    return None

print({'python': sys.version, 'platform': platform.platform(),
       'cpu_count': os.cpu_count(), 'memory_gib': memory_gib(),
       'free_disk_gib': shutil.disk_usage('/kaggle/working').free / (1024 ** 3),
       'kaggle_kernel_run_type': os.environ.get('KAGGLE_KERNEL_RUN_TYPE', 'unknown'),
       'execution_policy': 'CPU-only; workers_per_model=1; no benchmark started'})
```
